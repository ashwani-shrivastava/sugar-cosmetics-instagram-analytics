```python
# Check exact duplicate rows
duplicate_count = df.duplicated().sum()
duplicate_count
```

```python
# Select numerical columns
numeric_columns = df.select_dtypes(include="number").columns
# Count zero values
print("Zero Values in Each Numerical Column:")
print((df[numeric_columns] == 0).sum())
# Count negative values
print("\nNegative Values in Each Numerical Column:")
print((df[numeric_columns] < 0).sum())
```

```python
# Check whether Reach exceeds Impressions
invalid_reach = df[df["Reach"] > df["Impressions"]]
print("Records Where Reach Exceeds Impressions:", len(invalid_reach))
```

```python
# Inspect unique categorical values
categorical_columns = [
    "Post Type",
    "Campaign Type",
    "Audience Emotion",
    "Engagement Source"
]

for col in categorical_columns:
    print(f"\nUnique values in {col}:")
    print(df[col].value_counts())
```

```python
# Display the first 10 hashtag entries
df["Hashtags"].head(10)
```

```python
# Count the number of hashtags in each record
hashtag_count = (
    df["Hashtags"]
    .fillna("")
    .apply(lambda x: len([tag for tag in x.split(",") if tag.strip()]))
)

print("Minimum hashtags per record:", hashtag_count.min())
print("Maximum hashtags per record:", hashtag_count.max())
print("Average hashtags per record:", round(hashtag_count.mean(), 2))
```

```python
# Check the earliest and latest dates
print("Earliest Post Date:", df["Post Date"].min())
print("Latest Post Date:", df["Post Date"].max())
# Check for dates outside the expected project period
start_date = pd.Timestamp("2024-04-19")
end_date = pd.Timestamp("2025-04-19")

invalid_dates = df[
    (df["Post Date"] < start_date) |
    (df["Post Date"] > end_date)
]
print("Records Outside Expected Date Range:", len(invalid_dates))

# Check for missing dates
print("Missing Post Dates:", df["Post Date"].isna().sum())
```

```python
# Check for leading/trailing spaces and inconsistent capitalization

text_columns = [
    "Post Type",
    "Campaign Type",
    "Audience Emotion",
    "Engagement Source"
]

for col in text_columns:
    print(f"\nColumn: {col}")
    print("Unique values:", df[col].nunique())

    # Check values containing leading or trailing spaces
    spaces = df[col].astype(str).ne(df[col].astype(str).str.strip()).sum()
    print("Values with extra spaces:", spaces)

    # Display the unique values
    print(df[col].value_counts())
```

```python
# Check missing hashtags
print("Missing hashtags:", df["Hashtags"].isna().sum())

# Check empty or whitespace-only hashtags
empty_hashtags = df["Hashtags"].astype(str).str.strip().eq("").sum()

print("Empty hashtags:", empty_hashtags)

# Display unique hashtags strings
print("Unique hashtag combinations:", df["Hashtags"].nunique())
```

```python
# Check for negative values in numerical columns
numeric_columns = df.select_dtypes(include="number").columns

negative_counts = (df[numeric_columns] < 0).sum()

print("Negative values by column:")
print(negative_counts)

print("\nTotal negative values:", negative_counts.sum())
```
